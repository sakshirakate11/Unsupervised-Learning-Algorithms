 step 1: Import Libraries

```python
import numpy as np
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv("/content/sample_data/Groceries_dataset.csv")
df.head()
```

```python
basket = df.groupby(['Member_number', 'Date'])['itemDescription'].apply(list).reset_index()
transactions = basket['itemDescription'].tolist()

print(transactions[:5])
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df_encoded = pd.DataFrame(te_array, columns=te.columns_)
df_encoded.head()
```

```python
frequent_itemsets = apriori(df_encoded, min_support=0.01, use_colnames=True)

print("Total Frequent Itemsets:", frequent_itemsets.shape[0])
frequent_itemsets.head()
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.1)

rules = rules[
    (rules['antecedents'].apply(lambda x: len(x) >= 1)) &
    (rules['consequents'].apply(lambda x: len(x) >= 1))
]

print("Total Rules:", rules.shape[0])

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head()
```
